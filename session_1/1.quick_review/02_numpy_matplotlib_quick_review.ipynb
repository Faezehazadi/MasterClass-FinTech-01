# NumPy + Matplotlib: see the data

A quick review of data visualization with `Matplotlib`.

Numbers are useful; pictures help us notice patterns. 

We will use `NumPy` to prepare arrays and `Matplotlib` to make a bar chart, scatter plot, and line plot.

Topics: Arrays, vectorized operations, basic statistics, and plotting.

> **Estimated time: 10 min** · A quick review for a data science session


```python
import numpy as np
import matplotlib.pyplot as plt


# Make plots appear inside the notebook
%matplotlib inline
```

## 1 · Compare categories with a bar chart

Bar charts are a good choice when each label represents a category.

```python
students = np.array(['Adele', 'Fateme', 'Faeze', 'Maryam', "Faezeh"])
scores = np.array([79, 91.5, 84.75, 96, 79])

plt.bar(students, scores)
plt.title('Quiz scores')
plt.ylabel('Score')
plt.show()
```

### Scatter Plot

A scatter plot shows how two numerical measurements move together.

```python
apples_width = np.random.normal(6, 0.5, 100)
apples_length = np.random.normal(6, 0.5, 100)
bananas_width = np.random.normal(4, 0.5, 100)
bananas_length = np.random.normal(8, 0.5, 100)

plt.figure(figsize=(6, 4))
plt.scatter(apples_width, apples_length, label='Apple', color='crimson')
plt.scatter(bananas_width, bananas_length, label='Banana', color='gold')
plt.title('Fruit measurements')
plt.xlabel('Width')
plt.ylabel('Length')
plt.legend()
plt.show()
```

### Bar chart with error bars

A bar chart can show the mean of each category, with error bars to indicate the standard deviation.

```python
labels = ['bananas', 'apples', 'bananas', 'apples']

x = [1, 1.5, 2.5, 3.0]

height = [bananas_width.mean(),apples_width.mean(),bananas_length.mean(),apples_length.mean()]
yerr = [bananas_width.std(),apples_width.std(),bananas_length.std(),apples_length.std()]

color = ['gold', 'crimson', 'gold', 'crimson']

plt.bar(x = x,
        height=height,
        color=color,
        width=0.4,
        error_kw={'lw': 4},
        yerr=yerr
)

plt.xticks(ticks=x, labels=labels)
plt.title('Fruit length VS width')

plt.grid()
```

### Box diagram

A box diagram (or box plot) shows the median, quartiles, and outliers of a distribution. It is a good choice when you want to compare distributions across categories.

```python
plt.boxplot(
    bananas_length,
    patch_artist=True, 
    boxprops={'facecolor': 'gold'},
    positions=[0], 
    tick_labels=['bananas'], 
    widths=[0.15]
)

plt.boxplot(
    apples_length,
    patch_artist=True, 
    boxprops={'facecolor': 'crimson'},
    positions=[0.3], 
    tick_labels=['apples'], 
    widths=[0.15]
)

plt.ylabel('length')
plt.xlim(-0.2, 0.5)
plt.title('Length Distribution ')
```

### Histogram

A histogram shows the distribution of a single numerical measurement. It is a good choice when you want to see the shape of a distribution.

```python
plt.hist(bananas_width,
         bins=30, density=True, color='gold', label='bananas')

plt.hist(apples_width,
         bins=30, density=True, color='crimson', label='apples', alpha=0.7)

plt.legend(frameon=False)
plt.title('Width Distribution ')
plt.xlabel('Width')
plt.show()
```

## 3 · Track change with a line plot

Line charts are useful when the x-axis has an order, such as time.

```python
months = np.arange(1, 8)

month_names = [
    "فروردین", "اردیبهشت", "خرداد",
    "تیر", "مرداد", "شهریور", "مهر"
]

# میانگین ماهانه‌ی بازار آزاد
dollar_price = np.array([
    156335,  # فروردین
    173621,  # اردیبهشت
    170615,  # خرداد
    179146,  # تیر
    189113,  # مرداد
    221120,  # شهریور
    256496   # مهر
])

plt.plot(months, dollar_price, marker="o", linewidth=2, color="seagreen")
plt.title("قیمت دلار آمریکا در ایران - سال ۱۴۰۵")
plt.xlabel("ماه")
plt.ylabel("قیمت دلار (تومان)")
plt.xticks(months, month_names, rotation=30)
plt.grid(alpha=0.3)
plt.show()
```

## Key takeaways
- `bar` compares categories 📊.
- `scatter` explores relationships between two measurements.
- `plot` shows ordered change, often over time 📉📈.
- Titles, labels, limits, and legends make a chart understandable.
