# NumPy: numerical operations

A quick review of numerical operations with `NumPy`.

NumPy gives Python fast, compact arrays for numerical data. A famous library that we will use on everything.

**Topics:** Arrays, vectorized operations, basic statistics.

> **Estimated time: 10 min**

```python
import numpy as np
```

## 1. NumPy Arrays  

NumPy provides efficient arrays and vectorized operations for numerical computing.

```python
guys = np.array(["Arian", "AmirHossein", "Arian", "Ali", "Sajjad", "Saeed", "MohammadMahdi", "Mehrdad", "Yasin"])
ages = np.array([23, 26, 24, 23, 27, 31, 22, 26, 21])

print('Guys:', guys)
print('Ages:', ages)
print('Shape:', ages.shape)
print('Data type:', ages.dtype)
```

## 2. Basic Statistics

NumPy provides convenient functions for summarizing numerical data.

```python
print('Mean age:', ages.mean())
print('Median age:', np.median(ages))
print('Youngest:', ages.min())
print('Oldest:', ages.max())
print('Standard deviation:', ages.std())

# Index of the oldest guy
oldest_idx = np.argmax(ages)

print("Oldest student:", guys[oldest_idx])
print("Age:", ages[oldest_idx])
```

## 3 · Vectorized operations

Apply an operation to every value at once—no `for` loop needed. This is one of NumPy's most useful ideas.

```python
ages_next_year = ages + 1
print('Next year:', ages_next_year)
```

```python
adult = ages >= 18
print('Adult:', adult)
print('All are adults:', adult.all())
```

## 4 · A small numerical model

Arrays can represent measurements. Here, estimate house prices from area using a simple rule.

```python
area_m2 = np.array([60, 75, 83, 112, 145, 170])
price = area_m2 * 20.5 + 30
price
```
